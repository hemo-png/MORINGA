### Exercise 23 — Student Performance Analysis
- upload the student file and answer the questions: average score, highest/lowest score, top student, course counts, average score by course. Then create a bar chart of average score by course, a histogram of scores, and a bar chart of students by course.

````python


# Your code here
students_data = pd.read_excel("Student_Performance_Analysis_Complex .xlsx", sheet_name="Sheet1")
students_data.head()



````

````python
students_data.info()
````

````python
students_data.describe()
````

````python
#finding missing values in the dataset
students_data.isna().sum()
````

````python
#Finding the duplicate values in the dataset
students_data.duplicated()
````

````python
#missing values in the dataset
students_cleaned = students_data.dropna(subset=["Internet_Access","Study_Hours_Per_Week","Attendance_Percentage","Previous_GPA","Quiz_Score","Participation_Score"])
````

````python
students_cleaned.isna().sum()
````

````python
students_cleaned.describe()
````

````python
#
students_cleaned["Student_Name"]=students_cleaned["Student_Name"].str.upper()
students_cleaned["Gender"]=students_cleaned["Gender"].str.upper()
students_cleaned["County"]=students_cleaned["County"].str.upper()
students_cleaned["Internet_Access"]=students_cleaned["Internet_Access"].str.upper()
students_cleaned["Learning_Mode"]=students_cleaned["Learning_Mode"].str.upper()
students_cleaned["Performance_Level"]=students_cleaned["Performance_Level"].str.upper()
students_cleaned["Pass_Status"]=students_cleaned["Pass_Status"].str.upper()
students_cleaned["Program"]=students_cleaned["Program"].str.upper()
````

````python
students_cleaned.head()
````

````python
#Average scores for each program
average_scores = students_cleaned.groupby("Program")["Quiz_Score"].mean()
average_scores
````

````python
#average score bar graph
average_scores.plot(kind="bar", title="Average Quiz Scores by Program", xlabel="Program", ylabel="Average Quiz Score")
````

````python
#Highest and lowest scores for each program
highest_scores = students_cleaned.groupby("Program")["Quiz_Score"].max()
lowest_scores = students_cleaned.groupby("Program")["Quiz_Score"].min()
highest_scores, lowest_scores

````

````python
#Top students based on Overall_Score
top_students = students_cleaned.sort_values(by="Overall_Score", ascending=False).head()
#top_students= students_cleaned.groupby
top_students
````

````python
#Course counts
course_counts = students_cleaned["Program"].value_counts()
course_counts
````

````python
#Create a bar graph for the course by students
course_counts.plot(kind="bar", title=" Students in different programs", xlabel="Program", ylabel="Number of Students ", )
````

````python
# Create a histogram for the Overall_Score
students_cleaned["Overall_Score"].plot(kind="hist", title="Distribution of Overall Scores", xlabel="Overall Score", ylabel="Frequency", bins=10, color="blue", edgecolor="black")
````

### Exercise 24 — Sales Data Analysis
Create Revenue = Units_Sold × Price. Analyze total units, total revenue, average revenue, highest-selling product, highest-revenue product, revenue by region/product, average units by region, and sorted revenue. Create four requested visualizations and write 3–5 observations.

````python
data = {
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone", "Monitor", "Tablet", "Laptop", "Phone", "Monitor"],
    "Region": ["Nairobi", "Mombasa", "Nairobi", "Kisumu", "Nairobi", "Mombasa", "Nairobi", "Kisumu", "Mombasa", "Nairobi"],
    "Units_Sold": [10, 25, 15, 12, 30, 8, 20, 18, 22, 10],
    "Price": [80000, 50000, 80000, 30000, 50000, 25000, 30000, 80000, 50000, 25000]
}
df = pd.DataFrame(data)

# Your code here
Revenue = df["Units_Sold"] * df["Price"]
total_revenue = Revenue.sum()
total_revenue

````

````python
total_units_sold = df['Units_Sold'].sum()
total_units_sold
````

````python
#average revenue 
average_revenue = df["Price"].mean()
print(f"The average revenue is: {average_revenue}")
````

````python
#Highest and lowest selling products
product_revenue = df.groupby("Product")["Units_Sold"].sum()
top_product_revenue = product_revenue.idxmax()
highest_revenue = product_revenue.max()
print(f"The highest selling product is: {top_product_revenue} with {highest_revenue} units sold.")

````

````python
#highest revenue product
highest_revenue_product = df.groupby("Product")["Units_Sold"].sum().idxmax()

print(f"The product with the highest revenue is: {highest_revenue_product}")
````

````python
#revenue per region 
#revenue_region = df.groupby(["Region", "Product","Units_Sold", "Price"]).sum().reset_index()
#print(f"Revenue per region:\n{revenue_region}")
#print("---Total revenue breakdown---")
#print(revenue_region)
````

````python
#revenue per region
df["Revenue"]=df["Units_Sold"] * df["Price"]
total_units = df["Units_Sold"].sum()
total_revenue = df["Revenue"].sum()
pivot_table = df.pivot_table(values="Revenue", index="Region", aggfunc="sum")
pivot_table
````

````python
#Average units sold per region
avg_units_region = df.groupby("Region")["Units_Sold"].mean()
avg_units_region
````

````python
#sorted revenue
sorted_revenue = df.sort_values("Revenue", ascending=False)
sorted_revenue
````

````python
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
````

````python
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))
````

````python

#total revenue by product bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Product",
    y="Revenue",
    hue="Product",
    estimator="sum",
    errorbar=None,
    color="red"
)

plt.xlabel("Product")
plt.ylabel("Total Revenue ")
plt.title("Total Revenue by Product")

plt.tight_layout()
plt.show()
````

````python
#Revenue per region bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Region",
    y="Revenue",
    hue="Region",
    estimator="sum",
    errorbar=None,
    color="Blue"
)

plt.xlabel("Region")
plt.ylabel("Total Revenue ")
plt.title("Total Revenue by Region")

plt.tight_layout()
plt.show()

````

````python
#Average units sold per region bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Region",
    y="Units_Sold",
    hue="Region",
    estimator="mean",
    errorbar=None,
    color=("red")
)

plt.xlabel("Region")
plt.ylabel("Average Units Sold")
plt.title("Average Units Sold by Region")

plt.tight_layout()
plt.show()
````

````python
#units sold by product and region bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="Region",
    y="Units_Sold",
    hue="Product",
    estimator="sum",
    errorbar=None,
    palette=["blue", "green", "orange", "red"]
)

plt.xlabel("Region")
plt.ylabel("Total Units Sold")
plt.title("Total Units Sold by Product and Region")

plt.legend(title="Product")

plt.tight_layout()
plt.show()
````

3. Total Revenue was (8,700,000)

5. No laptops sold in Mombasa and No monitors in Kisumu.
