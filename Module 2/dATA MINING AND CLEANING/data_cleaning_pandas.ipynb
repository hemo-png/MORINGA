**what is data cleaning?**</br>Data cleaning is the process of identfying and addressing problems that makes data unreliable/difficult to analyse
comomon mistakes:
common data cleaning problems include:

-Inconsistent Text

-Incorrect value

-Invalid values



An outlier which is an unusually high or low value compared to the observation is not automatically an error. For example, if the patient wait time is normally between 10 mins to 25 mins but there is a recorded value of a 90 mins wait time, this is clearly an outlier but that does not mean its erroneous. We will end up deciding whether to drop or retain the value\outlier. However, if negative age value violates the meaning of that column denoting that is erroneous.

**Data Cleaning Process**
-Preserve the raw data and create a working copy

-Inspect the data cleaning issues:missing values,duplicate values,erroneus values, column inconsistences,null values etc

-Deal with the issues:dropping null values,imputations, introduce uniformity etc

-Saved the cleaned copy

```python
# convert data types for the following rows: fees(int) and dates (datetime)


# 1. Convert visit_date to datetime
clinic_visits_copy['visit_date'] = pd.to_datetime(
    clinic_visits_copy['visit_date'], 
    errors='coerce', 
    format='mixed'
)

# 2. Clean fee_kes (remove symbols, text, decimals) and convert to integer
clinic_visits_copy['fee_kes'] = (
    clinic_visits_copy['fee_kes']
    .astype(str)
    .str.replace(r'[^\d.]', '', regex=True)  # Remove currency symbols/commas
)

# Convert to float first (to handle decimals like 1200.0), then to integer
clinic_visits_copy['fee_kes'] = (
    pd.to_numeric(clinic_visits_copy['fee_kes'], errors='coerce')
    .fillna(0)             # Or dropna/impute before integer conversion if needed
    .astype('int64')
)

# Verify updated datatypes
print(clinic_visits_copy[['visit_date', 'fee_kes']].dtypes)
```

```python
#step 7 : check the invalid ages and missing times


# 1. Check for invalid age values (negative ages or unrealistic ages over 120)
invalid_ages = clinic_visits_copy[
    (clinic_visits_copy['age'] < 0) | (clinic_visits_copy['age'] > 120)
]

print(invalid_ages[['visit_id', 'patient_id', 'age']])


# 2. Check for missing (null) values in wait_minutes
missing_wait_times = clinic_visits_copy[clinic_visits_copy['wait_minutes'].isnull()]

print(missing_wait_times[['visit_id', 'patient_id', 'wait_minutes']])


# 3. Check for invalid (negative) wait times
invalid_wait_times = clinic_visits_copy[clinic_visits_copy['wait_minutes'] < 0]

print(invalid_wait_times[['visit_id', 'patient_id', 'wait_minutes']])


# 4. Check for extreme wait time outliers using IQR (Interquartile Range)
Q1 = clinic_visits_copy['wait_minutes'].quantile(0.25)
Q3 = clinic_visits_copy['wait_minutes'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR

outlier_wait_times = clinic_visits_copy[clinic_visits_copy['wait_minutes'] > upper_bound]

print(outlier_wait_times[['visit_id', 'service', 'wait_minutes']])
```

"""
# Step 8: Handling Invalid, Missing, and Extreme Values

After identifying data-quality problems, we need to decide how to handle each
problem appropriately. The three main approaches are imputing, retaining, and
dropping values.

# 1. IMPUTING MISSING VALUES

When a value is missing, we can sometimes estimate a reasonable replacement
instead of deleting the entire record.

For missing wait times, we calculate the median wait time and use it to fill
the missing values.

The median is the middle value when all observations are arranged from
smallest to largest. It is useful for wait times because it is less affected
by extremely long waiting times than the mean.

For example, if the wait times are:

    10, 12, 15, 20, 25, 100

The value 100 is unusually high and could pull the average upward. The median
gives a better representation of a typical waiting time.

We therefore use:

    median_wait = clinic_visits_copy['wait_minutes'].median()

and then fill missing wait times using:

    clinic_visits_copy['wait_minutes'].fillna(median_wait)

This allows us to keep the patient record while providing a reasonable value
for the missing wait time.


# 2. IDENTIFYING EXTREME WAIT TIMES USING THE INTERQUARTILE RANGE (IQR)

Before deciding whether an unusually high wait time should be removed, we
first need to determine whether it is an extreme value.

The Interquartile Range (IQR) measures the spread of the middle 50% of the
data.

First, we calculate:

    Q1 = 25th percentile
    Q3 = 75th percentile

Then:

    IQR = Q3 - Q1

The upper boundary for potential outliers is calculated as:

    Upper Bound = Q3 + (1.5 × IQR)

Any wait time above this boundary is considered a potential outlier.

It is important to remember that an outlier is not automatically an error.
For example, a patient may genuinely have waited several hours. Therefore,
extreme wait times should be investigated before deciding whether to remove
them.

If an extreme value is legitimate, we RETAIN it.


# 3. DROPPING INVALID VALUES

Some values are clearly invalid and cannot reasonably be used.

For example:

    age = -5
    wait_minutes = -10

A negative age or negative waiting time does not make logical sense.

If the correct value cannot be determined from another reliable source, the
affected record can be removed from the dataset.

For example, we can keep only ages between 0 and 120 and only non-negative
wait times.

Dropping should be used carefully because removing records reduces the amount
of available data.


# 4. RETAINING VALID BUT UNUSUAL VALUES

Not every unusual value should be removed.

If an outlier is unusual but represents a real observation, we retain it.

For example, if a patient genuinely waited 150 minutes, the value may be an
outlier but it is not necessarily incorrect.

The goal of data cleaning is not to make every value look normal. The goal is
to make the dataset accurate and reliable while preserving legitimate
information.


# OVERALL DECISION PROCESS

The cleaning process can therefore be summarized as:

    Missing value
        ↓
    Can it be reasonably estimated?
        ↓
    Yes → Impute using an appropriate value such as the median

    Invalid value
        ↓
    Can the correct value be determined?
        ↓
    No → Drop the affected record

    Unusual/extreme value
        ↓
    Is it a genuine observation?
        ↓
    Yes → Retain it
    No  → Remove or correct it

The main principle is to choose the treatment based on the nature of the
problem rather than automatically deleting unusual or missing data.
"""

```python

#  INVALID VALUES & IMPUTATION


# 1. Fix negative values in ID columns (patient_id, visit_id, clinic_id)
id_cols = [col for col in ['patient_id', 'visit_id', 'clinic_id'] if col in clinic_visits_copy.columns]
for col in id_cols:
    clinic_visits_copy[col] = clinic_visits_copy[col].astype(str).str.replace('-', '', regex=False)

# 2. Impute missing visit_date (e.g., P10 / V011) using forward-fill
clinic_visits_copy['visit_date'] = clinic_visits_copy['visit_date'].ffill()

# 3. Handle invalid Age values (< 0 or > 120)
invalid_age_mask = (clinic_visits_copy['age'] < 0) | (clinic_visits_copy['age'] > 120)
clinic_visits_copy.loc[invalid_age_mask, 'age'] = pd.NA

median_age = clinic_visits_copy['age'].median()
clinic_visits_copy['age'] = clinic_visits_copy['age'].fillna(median_age)

# 4. Handle invalid negative Wait Times
clinic_visits_copy.loc[clinic_visits_copy['wait_minutes'] < 0, 'wait_minutes'] = pd.NA

# 5. Impute missing/invalid Wait Times by service category
clinic_visits_copy['wait_minutes'] = clinic_visits_copy.groupby('service')['wait_minutes'].transform(
    lambda x: x.fillna(x.median())
)
clinic_visits_copy['wait_minutes'] = clinic_visits_copy['wait_minutes'].fillna(
    clinic_visits_copy['wait_minutes'].median()
)

# 6. Handle fee_kes missing values
if 'fee_kes' in clinic_visits_copy.columns:
    clinic_visits_copy['fee_kes'] = clinic_visits_copy.groupby('service')['fee_kes'].transform(
        lambda x: x.fillna(x.median())
    )
```

```python
#step 9:validate the cleaned Data and export(save) it as a CSV file

# 1. Inspect data structure and data types

print(clinic_visits_copy.info())

# 2. Verify zero remaining missing (null) values

print(clinic_visits_copy.isnull().sum())

# 3. Verify zero remaining duplicate rows

print(f"Duplicate row count: {clinic_visits_copy.duplicated().sum()}")

# 4. Statistical summary to confirm ranges (age, fee_kes, wait_minutes)

print(clinic_visits_copy.describe())

# 5. Preview first few clean records

print(clinic_visits_copy.head())


# EXPORT TO CSV


output_filename = "clinic_visits_cleaned.csv"
clinic_visits_copy.to_csv(output_filename, index=False)

print(f"\n Validation complete! Cleaned dataset successfully saved as '{output_filename}'.")
```
