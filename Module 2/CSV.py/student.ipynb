```python
import csv
students = [
    ['Mary',30,'Nakuru'],
    ['Job',27,'Mombasa'],
    ['Bob',34,'Meru']
]

with open('students.csv','w',newline="") as file:
    writer = csv.writer(file)
    writer.writerow(['Name','Age','Town'])#header
    writer.writerows(students)#rows

for x in students:
    print(x)
```

```python
#append new student
students = ['Maximila',33,'Margret']
with open('students.csv','a',newline='')as f:
    writer= csv.writer(f)
    writer.writerow(students)
```

```python
students = ['Charles',39,'Makueni']
with open('students.csv','a',newline='')as f:
    writer= csv.writer(f)
    writer.writerow(students)
```

```python
#reading Csv content
with open ('students.csv','r') as file:
    reader = csv.reader(file)

    for x in reader:
        print(x)
```

```python
#diict reader
with open ('students.csv','r') as file:
    reader = csv.DictReader(file)

    for x in reader:
        print(x)
```
