#Data cleansing with pandas
**What is data cleansing?**</br>Data cleaning is the process of identiying and addresing problems that make data unreliable/difficult to analyse.
Common data cleaning problems include:
-missing values
-duplicate values
-incostitent values
-incorrect data types
-invalid values

An outlier which is unusully high or low value compared to the observation is not automatically an error. for example. if the patient wait time is normally between 10min to 15 min but there is arecord value of 50 min wait time, this is clearly an outlier but that does not mean its erroneous . we will end up deciding wether to drop or retain the values/outlier However, a negative age violates the meaning o the meaning of that column denoting that its erroneous.

**Data Cleansing Process**
-presreve the raw data and create a working copy.
-inspect the Data cleansing issues:missing values, duplicate values, erroneous values, column inconsistencies, null values e.t.c
-Deal with the issues example dropping null values, imputations, introduce uniormity e.t.c
-save the cleaaned copy.

```python
#step 1 : Load Data
import pandas as pd
clinics_visits = pd.read_csv('clinic_visits_raw.csv')
clinics_visits.head()
```

```python
#step 2: Make a copy of the raw data
clinics_visits_copy = clinics_visits.copy()
clinics_visits_copy.head()
```

***Data Cleaning Steps***
1.Checking for duplicates and missing values
2.Drop  duplicates and drop/inpute missing values
3.Check and handle outliers

```python
#step 4 :inspecting the data for null/missing values
clinics_visits_copy.isnull().sum()
#diff btwn isnull() and isna() is that isnull() is a method of pandas DataFrame and Series objects, while isna() is a method of pandas Series objects. Both methods return a boolean mask indicating the presence of missing values in the data.
```

```python
#step 5: Checking for duplicates
clinics_visits_copy.duplicated().sum()
```

```python
#identify the duplicated row
duplicated_rows = clinics_visits_copy[clinics_visits_copy.duplicated(keep=False)]
duplicated_rows
```

```python
#remove the duplicated row
drop_duplicates = clinics_visits_copy.drop_duplicates().reset_index(drop=True)
print(drop_duplicates)  
```

```python
#step  5:standardize the columns (remove the white spaces and consistent naming )
clinics_visits_copy.columns = clinics_visits_copy.columns.str.strip().str.lower().str.replace(' ', '_')
clinics_visits_copy
```
