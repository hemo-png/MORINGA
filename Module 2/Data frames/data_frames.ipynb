```python
import pandas as pd
#panda series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively referred to as the index. A pandas Series can be created from a list, dictionary, or scalar value.  
age = pd.Series([22, 34, 27, 44, 19], name='Age'        )
age
```

```python
#create a series that stores temprature
temperature = pd.Series([30, 32, 28, 35, 31], index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'], name='Temperature')
temperature 
```

```python
**Data frames**</br>are two-dimensional labeled data structures with columns of potentially different types. You can think of them like a spreadsheet or SQL table, or a dict of Series objects. It is generally the most commonly used pandas object.
```

```python
students = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [20, 21, 19, 22],
    'Country': ['America', 'Brazil', 'Canada', 'Australia']
}
print(students)
#turn it into a data frame
df = pd.DataFrame(students)
print(df)
```

```python

#1 Reassign the indexes to 101, 102, 103, and 104.
df.index = [101, 102, 103, 104]
print(df)
```

```python
#Display the names and ages of the students who are older than 20.
df[df['Age'] > 20][['Name', 'Age']]
```
