***Question 3***</br>
Create this nested tuple:</br>

data = ((2, 5), (7, 3))</br>

Use a loop to calculate the sum of each inner tuple.</br>

Your output should follow this style:</br>
Row 1: 7 </br>
Row 2: 10

```python
# [ANSWER HERE]
data = ((2,5), (7,3))
for i in range(len(data)):
    print(f"Row {i + 1}:{sum(data[i])}")
```

### LISTS

***Question 1***</br>
Build and edit a shopping list. Create this list:</br>

food = ["apples", "oats"]</br>

Then:</br>

- Add "yoghurt" using .append().</br>
- Add "tea" and "bread" using .extend().</br>
- Display the first two items using slicing.</br>
- Display the final item using indexing.</br>

```python
# [ANSWER HERE]
food = ["apples", "oats"]
food.append("yoghurt")
print(food)
food.extend(["tea", "bread"])
print(food)
print(food[:2])
print(food[-1])
```

***Question 2***</br>
Convert text into a List. Start with:</br>

breakfast_text = "oats, banana, tea"</br>

Then: </br>

- Use .split() to create a list called breakfast.</br>
- Use len() to check how many items are in the list.</br>
- Create a list called lengths containing the length of each breakfast item using a list comprehension.</br>
- Display breakfast and lengths.</br>

```python
# [ANSWER HERE]
breakfast_text = str.split("oats, banana, tea")
print(breakfast_text)
print(len(breakfast_text))
lengths = [len(item)for item in breakfast_text]
print(breakfast_text)
print(lengths)
```

### TUPLES & LISTS

***Question 1***</br>
Create a tuple containing three lists:</br>

word_bank = (</br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;["river", "cloud", "forest"],</br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;["drifts", "shines", "whispers"],</br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;["quiet", "bright", "green"]</br>
)</br>

The lists represent nouns, verbs, and adjectives.</br>

Then:</br>

- Unpack the tuple into nouns, verbs, and adjectives.</br>
- Import Python's random module.</br>
- Randomly select one noun.</br>
- Randomly select one verb.</br>
- Randomly select one adjective.</br>
- Create and print a sentence such as:</br>

The quiet river whispers.</br>

The exact sentence can change each time the program runs.

```python
# [ANSWER HERE]
world_bank = (
["river", "cloud", "forest"],
["drifts", "shines", "whispers"],
["quiet", "bright", "green"]
)
nouns,verbs,adjectives = world_bank 
import random
nouns = random.choice(nouns)
verbs = random.choice(verbs)
adjectives = random.choice(adjectives)
print(f"{adjectives},{nouns},{verbs}")


```
