```python
import csv
sales = [
    ['Laptop',5,80000],
    ['Mouse',20,1500],
    ['Keyboard',10,3000],
    ['Monitor',7,25000]
]
with open('sales.csv','w',newline="") as file:
    writer = csv.writer(file)
    writer.writerow(['Product','Quantity','Price'])#header
    writer.writerows(sales)#rows

for x in sales:
    print(x)
```

```python
sales  = ['Cables',45,'24500']
with open('sales.csv','a',newline='')as f:
    writer= csv.writer(f)
    writer.writerow(sales)

```

```python
sales  = ['Stylus',23,'46500']
with open('sales.csv','a',newline='')as f:
    writer= csv.writer(f)
    writer.writerow(sales)
```

```python
with open('sales.csv','r',newline="") as file:
    reader = csv.DictReader(file)


    for x in reader:
        print(x['product'])
    
```
